## Stemming

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    l=[]

    for i in text.split():
        l.append(ps.stem(i))

    return " ".join(l)
    
```

```python
new_movie['tags']=new_movie['tags'].apply(stem)
```

```python
new_movie['tags'][0]
```

```python
#now we will vectorise it with the help of sklearn
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features= 5000, stop_words = 'english')
```

```python
vectors = cv.fit_transform(new_movie['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity = similarity.astype("float32")
```

## final function

```python
def recommend(movie):
    movie_index = new_movie[new_movie['title']==movie].index[0]      # this will find index of the user's movies
    distances = similarity[movie_index]                              # this will find similarity of user's movies to other movies
    movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]      # enumerate will create all data with its index in tuple format so that after sorting ccorresonding index will not lost

    for i in movies_list:
        print(new_movie.iloc[i[0]].title)                            # 
```
